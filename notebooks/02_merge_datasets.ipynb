## Merge Datasets


```python
import pandas as pd
from pathlib import Path
```

```python
ROOT_DIR = Path("../data/interim")
DATA_CONFIG = {
    "krishna_rainfall" : ("krishna_rainfall.csv"),
    "krishna_streamflow" : ("krishna_streamflow.csv"),
    "krishna_tmax" : ("krishna_tmax.csv"),
    "krishna_tmin" : ("krishna_tmin.csv"),
    "krishna_temperature" : ("krishna_temperature.csv"),

    "kaveri_rainfall" : ("kaveri_rainfall.csv"),
    "kaveri_streamflow" : ("kaveri_streamflow.csv"),
    "kaveri_tmax" : ("kaveri_tmax.csv"),
    "kaveri_tmin" : ("kaveri_tmin.csv"),
    "kaveri_temperature" : ("kaveri_temperature.csv"),

    "oni" : ("oni.csv"),
}

datasets = {}

for name, filename in DATA_CONFIG.items():
    datasets[name] = pd.read_csv(ROOT_DIC / filename)
```

```python
krishna_weather = (
    datasets["krishna_rainfall"]
        .merge(datasets["krishna_temperature"], on="date")
        .merge(datasets["oni"], on="date", how="left")
)

krishna_weather.to_csv(
    ROOT_DIR / "krishna_weather.csv",
    index=False
)
```

```python
kaveri_weather = (
    datasets["kaveri_rainfall"]
        .merge(datasets["kaveri_temperature"], on="date")
        .merge(datasets["oni"], on="date", how="left")
)

kaveri_weather.to_csv(
    ROOT_DIR / "kaveri_weather.csv",
    index=False
)
```

```python
krishna_master_dataset = krishna_weather.merge(datasets["krishna_streamflow"])
krishna_master_dataset.to_csv("../data/processed/krishna_master_dataset.csv", index=False)
```

```python
kaveri_master_dataset = kaveri_weather.merge(datasets["kaveri_streamflow"])
kaveri_master_dataset.to_csv("../data/processed/kaveri_master_dataset.csv", index=False)
```
